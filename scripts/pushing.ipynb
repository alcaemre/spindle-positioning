```python
import src.spindle_state as ss

import numpy as np
# np.set_printoptions(formatter={'float': '{:.3f}'.format})
import matplotlib.pyplot as plt
%matplotlib widget

from rich.console import Console
from rich.live import Live
from rich.table import Table

console = Console()
```

## 1-D

Looking for:
- stable fixed point at $0$ when `critical_length` $ < 1$,
-  degenerate force balance when `critical_length` $ > 1$,

```python
test_1d_lattice = np.array([
    [1, 0, 0],
    [-1, 0, 0],
])
test_1d_spindle_state = np.array([1,1]) # both pushing

initial_mtoc_pos_1d = np.array([0.6,0,0])

stall_force = 1
critical_length = 0.5
rigidity = stall_force * np.square(critical_length) / np.square(np.pi) 

test_spindle_1d = ss.Spindle(
    initial_mtoc_pos_1d, 
    test_1d_spindle_state, 
    test_1d_lattice, 
    timestep_size=0.001,
    stall_force=stall_force,
    rigidity=rigidity,
    )

test_spindle_1d.add_microtubules(np.array([0,1]))

test_spindle_1d.calculate_pushing_forces()

# # -- relax --
with Live(console=console, refresh_per_second=4) as live:
    traj, boundary_violated = test_spindle_1d.relax(readout=True, live=live)

    test_spindle_1d.trajectory = test_spindle_1d.trajectory | traj
    test_spindle_1d.current_time = list(traj.keys())[-1]

ani = test_spindle_1d.animate(interval=1000)

plt.show()
```

## 2-D
Looking for: 
- stable fixed point at $\vec{0}$ when `critical_length` $ < 1$,
- unstable fixed point at $\vec{0}$ when `critical_length` $ > 1$,

```python
test_2d_lattice = np.array([
    [1, 1, 0],
    [1, -1, 0],
    [-1, 1, 0],
    [-1, -1, 0],
])
test_2d_spindle_state = np.array([1,1,1,1]) # both pushing

initial_mtoc_pos_2d = np.array([0.3,0.2,0])

# setting critical length and consequently rigidity
critical_length = 0.5 * np.sqrt(2)
stall_force = 1
rigidity = stall_force * np.square(critical_length) / np.square(np.pi) 

test_spindle_2d = ss.Spindle(
    initial_mtoc_pos_2d, 
    test_2d_spindle_state, 
    test_2d_lattice, 
    timestep_size=0.001,
    stall_force=stall_force,
    rigidity=rigidity,
    )

test_spindle_2d.add_microtubules(np.array([0,1,2,3]))

test_spindle_2d.calculate_pushing_forces()

# -- relax --
with Live(console=console, refresh_per_second=4) as live:
    traj, boundary_violated = test_spindle_2d.relax(readout=True, live=live)

    test_spindle_2d.trajectory = test_spindle_2d.trajectory | traj
    test_spindle_2d.current_time = list(traj.keys())[-1]

ani = test_spindle_2d.animate(interval=1000)

plt.show()
```
