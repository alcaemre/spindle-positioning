```python
def restart_experiment_from_directory(dir_path, num_rounds):
    """restarts an experiment from a particular dictionary

    Args:
        dir_path (str): path to experiment directory
        num_rounds (int): number of stable positions found to stop at

    Returns:
        Spindle: spindle upon finishing simulation with that number of rounds
    """

    # load spindle initialization data
    spindle_path = os.path.join(dir_path, 'spindle.pkl')
    with open(spindle_path, 'rb') as f:
        spindle_dict = pickle.load(f)

    # load spindle trace
    trace_path = os.path.join(dir_path, 'spindle_trace.pkl')
    with open(trace_path, "rb") as f:
        spindle_trace = pickle.load(f)
    # initialize spindle
    spindle_from_dict = ss.init_spindle_from_dict(spindle_dict)


    # pick up where trace leaves off
    spindle_from_dict.spindle_trace = spindle_trace
    last_spindle_trace = spindle_trace[list(spindle_trace.keys())[-1]]

    spindle_from_dict.set_mtoc_pos(last_spindle_trace['mtoc_pos'])
    spindle_from_dict.spindle_state = last_spindle_trace['spindle_state']
    spindle_from_dict.current_time = list(spindle_trace.keys())[-1]

    # simulate until we have num_rounds stable positions found

    spindle_from_dict.single_mt_update_simulation_to_equilibrium(num_rounds, initial_cost=last_spindle_trace['cost'],  save=True, readout=True)

    return spindle_from_dict

dir_path = '/Users/emrealca/Documents/Penn/flatiron-microtubules/simulations/data/2D-single-mt-update-100-points-10-mts-3-positions_2026-02-20_17-09-05'

restart_experiment_from_directory(dir_path, 10)
```
